```python
from ultralytics import YOLO
from PIL import Image
```

```python
model = YOLO('yolov8x-worldv2.pt')
custom_classes = ["duck"]
image = Image.open("/nas_data/dataset/4dhoi/25_04_03/person00/motion07/images/16/ori_imgs/000006.jpg")
model.set_classes(custom_classes)
results = model.predict(image,conf=0.001)
result = results[0]
for i, box in enumerate(result.boxes):
    # 获取边界框坐标 (xyxy格式: x_min, y_min, x_max, y_max)
    xyxy = box.xyxy[0].cpu().numpy()
    x_min, y_min, x_max, y_max = xyxy

    # 获取置信度
    confidence = box.conf[0].cpu().numpy()

    # 获取类别ID (这是对应到你 custom_classes 列表的索引)
    class_id = int(box.cls[0].cpu().numpy())

    # 从你的自定义类别列表中获取类别名称
    class_name = custom_classes[class_id]

    print(f"  物体 {i+1}:")
    print(f"    类别: {class_name}")
    print(f"    置信度: {confidence:.4f}") # <--- 这里是边界框的置信度
    print(f"    边界框 (xyxy): [{x_min:.2f}, {y_min:.2f}, {x_max:.2f}, {y_max:.2f}]")
result.save(filename='result_yolo_world.jpg')
```
